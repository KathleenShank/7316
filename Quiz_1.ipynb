**10. Plot the image.**
This is a tricky one because the image (as you see above) is huge. In fact, it is too big to fit in Google Colab's memory. So...read a `rasterio.windows.Window` of the image. See the documentation. I suggest a window not larger than 3000 x 2000. BUT, you can use AI for this question. You can use `matplotlib` or `ee` and `geemap`.

````python
# DEFINE THE SIZE OF THE PLOT
# Max window size
MAX_WIN_W = 3000   # columns (width)
MAX_WIN_H = 2000   # rows (height)

# Output display size
DISPLAY_W = 1500   # width to display (pixels)
DISPLAY_H = 1000   # height to display (pixels)
````

````python
# Using mapplotlib #
# DEFINE THE WINDOW FOR THE PLOT BASED ON ABOVE SIZE VARIABLES
def read_window_for_plot(src, win_col_off=None, win_row_off=None,
                         win_w=MAX_WIN_W, win_h=MAX_WIN_H,
                         out_w=DISPLAY_W, out_h=DISPLAY_H):
    """Read a window (with bounds-checking) and resample to out_shape for plotting.
       Returns numpy array with shape (rows, cols, bands) suitable for imshow.
    """
    # ensure integer sizes
    win_w = int(min(win_w, src.width))
    win_h = int(min(win_h, src.height))

    # default: center the window
    if win_col_off is None:
        win_col_off = max(0, (src.width - win_w) // 2)
    if win_row_off is None:
        win_row_off = max(0, (src.height - win_h) // 2)

    # ensure window is within raster bounds
    if win_col_off + win_w > src.width:
        win_col_off = src.width - win_w
    if win_row_off + win_h > src.height:
        win_row_off = src.height - win_h

    window = Window(win_col_off, win_row_off, win_w, win_h)

    # Read all bands, but request an out_shape to resample to smaller display size
    # out_shape expects (bands, rows, cols)
    out_shape = (src.count, out_h, out_w)

    data = src.read(window=window, out_shape=out_shape, resampling=rasterio.enums.Resampling.bilinear)
    # data shape: (bands, rows, cols)
    # Convert to (rows, cols, bands) for matplotlib
    data = np.transpose(data, (1, 2, 0))
    return data, window
````

````python
# SCALE IMAGE
def scale_for_display(img, percentiles=(2, 98)):
    """Scale image to uint8 for display using percentile stretch
       img: (rows,cols,bands) numpy array
       returns uint8 image of same shape (rows,cols,bands)
    """
    img_out = img.copy().astype('float32')
    bands = img_out.shape[2]

    for b in range(bands):
        band = img_out[..., b]
        vmin, vmax = np.percentile(band[~np.isnan(band)], percentiles) if np.any(~np.isnan(band)) else (np.nanmin(band), np.nanmax(band))
        if vmax == vmin:
            # avoid division by zero
            band_scaled = np.clip((band - vmin), 0, 1)
        else:
            band_scaled = (band - vmin) / (vmax - vmin)
        img_out[..., b] = np.clip(band_scaled, 0, 1) * 255.0

    return img_out.astype('uint8')
````

````python
# Open, read a window, and plot
# Path to file
filepath = "LaCimaA120190425Processed_Orthomosaic_export_SatMay11023009.304783.tif"

# Print everything defined above
with rasterio.open(filepath) as src:
    print("Full raster size (width, height):", src.width, src.height)
    print("Band count:", src.count)
    print("CRS:", src.crs)
    print("Bounds:", src.bounds)

    # Read a centered window (you can pass offsets to read a different region)
    img_window, window = read_window_for_plot(src)
    print("Read window:", window)
    print("Image chunk shape (rows, cols, bands):", img_window.shape)

# If the raster has >=3 bands we assume bands 1,2,3 are RGB; if >3 (e.g. 4 bands), we'll still use first 3
# If it's single-band, display it as grayscale
````

````python
# For loop for showing all the image accessories
if img_window.ndim == 3 and img_window.shape[2] >= 3:
    rgb = img_window[..., :3]
    rgb8 = scale_for_display(rgb)
    plt.figure(figsize=(10, 7))
    plt.imshow(rgb8)
    plt.title(f"Window (cols={int(window.width)}, rows={int(window.height)}) at col_off={int(window.col_off)}, row_off={int(window.row_off)}")
    plt.axis('off')
    plt.show()
else:
    # single-band
    band = img_window[..., 0] if img_window.ndim == 3 else img_window
    # scale to 0-255
    band8 = scale_for_display(band[..., None])[..., 0]
    plt.figure(figsize=(10, 7))
    plt.imshow(band8, cmap='gray')
    plt.title("Single-band window")
    plt.axis('off')
    plt.show()
````
